```python
from Knoten import Knoten
```

```python
level = [
    ["#", "#", "#", "#"],
    ["#", " ", "*", "#"],
    ["#", "*", "#", "#"],
    ["#", "#", "#", "#"]
]
```

```python
pacman_pos_x = 1
pacman_pos_y = 1
```

```python
if __name__ == "__main__":
    start = Knoten(level, pacman_pos_x, pacman_pos_y)

    offene_knoten = [start]
    zaehler = 0

    while zaehler < 10 and len(offene_knoten) > 0:
        aktueller_knoten = offene_knoten.pop(0)

        print("Expandierter Knoten:", zaehler + 1)

        for zeile in aktueller_knoten.level:
            print(zeile)

        print("Kinder:")
        kinder = aktueller_knoten.expand()

        for kind in kinder:
            for zeile in kind.level:
                print(zeile)
            print("---")

        offene_knoten.extend(kinder)

        print("====================")
        zaehler += 1

#Beim Expandieren entstehen Zyklen.

#Ein bereits besuchter Zustand kann erneut erzeugt werden.
#Dadurch wächst der Suchbaum unnötig stark an und dieselben
#Zustände werden mehrfach untersucht.

#Lösung:
#Eine Closed List (Visited-Liste) speichern und bereits
#besuchte Zustände nicht erneut expandieren.
```
