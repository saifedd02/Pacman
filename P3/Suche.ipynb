# Suchalgorithmen
Nutzt das Notebook für die erste Implementierung eurer Suchalgorithmen, um die Funktionalität zu testen. Kopiert die fertige Klasse dann direkt in [Suche.py](Suche.py).

**Tipp:** Erstellt euch ein statisches Numpy Array mit einem Startknoten und einem Zielknoten, um die Funktionalität eurer Suchalgorithmen zu testen, so müsst ihr nicht immer den Server starten.

```python
from Knoten import Knoten
from game_core.config import TILE_TYPES
```

```python
class Suche:

    def ist_ziel(self, knoten):

        for zeile in knoten.view:

            for feld in zeile:

                if feld == TILE_TYPES["*"]:
                    return False

        return True


    def starte_Suchalgorithmus(self, node: Knoten) -> Knoten:
        """
        node: Startknoten
        @return: Zielknoten oder None
        """

        open_list = [node]

        while len(open_list) > 0:

            aktueller_knoten = open_list.pop(0)

            if self.ist_ziel(aktueller_knoten):
                return aktueller_knoten

            kinder = aktueller_knoten.expand()

            for kind in kinder:
                open_list.append(kind)

        return None
```

```python
level = [
    [TILE_TYPES["#"], TILE_TYPES["#"], TILE_TYPES["#"], TILE_TYPES["#"]],
    [TILE_TYPES["#"], TILE_TYPES["P"], TILE_TYPES["*"], TILE_TYPES["#"]],
    [TILE_TYPES["#"], TILE_TYPES[" "], TILE_TYPES[" "], TILE_TYPES["#"]],
    [TILE_TYPES["#"], TILE_TYPES["#"], TILE_TYPES["#"], TILE_TYPES["#"]]
]

start = Knoten(level, (1, 1))

suche = Suche()
ziel = suche.starte_Suchalgorithmus(start)

print(ziel)
```
